```python
# Student ID- F415563
# In this file, we have worked on making a GUI for all the functions our codes have to display.
# We have started by importing all the modules and importing the data from other files to make this file run solely
# on the code that is required for it.
# Here we have created a GUI Graphical User Interface  in this notebook using which we will analyze the music data which we have stored in SQLite database named CWDatabase.db.
# Here we can explore the data like genre, artist and top5 artist in specific time duration.
# The main purpose is to display music genre, artist popularity and top 5 artist for each year between 1998-2020.
# We can choose any time frame between 1998-2020.
# For widgets we have used sliders, text boxes and buttons for better way to explore the data.
# The created interface has year slider, text box, start and end year slider, and buttons to get different results as per the input.
# In order to run this program we need to make sure we have all the files(genres.py,artist.py,Top5.py,Preprocessing.py and .db file) in same directory of jupyter notebook.
# Once done we need to import the .py files to process and view the data.
# We have used the callback function to connect with all the required database.
# Widgets helps us to input the data and do interaction.
# Layout which we have used basically help us to organize the widgets in proper manner.
# And lastly the output widget help us to get the desired result as per the data selection.

# %%

import ipywidgets as widgets
from IPython.display import display, clear_output
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt


from Artist import display_artist_statistics
from Genre import display_genre_statistics
from Top5 import TopArtistsAnalyzer


# output display 
output = widgets.Output()

# displaying genre stats
def on_create_genre_stats_clicked(b):
    with output:
        clear_output(wait=True)
        year = year_slider.value
        conn = sqlite3.connect("CWDatabase.db")
        display_genre_statistics(year, conn)
        conn.close()

# displaying artist popularity
def on_artist_popularity_clicked(b):
    with output:
        clear_output(wait=True)
        artist_name = artist_text.value
        conn = sqlite3.connect("CWDatabase.db")
        display_artist_statistics(artist_name, conn)
        conn.close()

# displaying top artists
def on_top_artists_clicked(b):
    with output:
        clear_output(wait=True)
        start_year = start_year_slider.value
        end_year = end_year_slider.value
        analyzer = TopArtistsAnalyzer("CWDatabase.db")
        data = analyzer.fetch_top_artists(start_year, end_year)
        analyzer.display_results(start_year, end_year, data)

# displaying reset (back to file loading)
def on_back_clicked(b):
    with output:
        clear_output(wait=True)
        print("Back to file loading...")

# GUI widgets
year_slider = widgets.IntSlider(value=2020, min=1998, max=2020, step=1, description='Year:')
artist_text = widgets.Text(placeholder='Enter artist name', description='Artist:')
start_year_slider = widgets.IntSlider(value=1998, min=1998, max=2020, step=1, description='Start Year:')
end_year_slider = widgets.IntSlider(value=2020, min=1998, max=2020, step=1, description='End Year:')

# buttons for the GUI
create_genre_stats_button = widgets.Button(description="Create Genre Stats for Year")
create_genre_stats_button.on_click(on_create_genre_stats_clicked)

artist_popularity_button = widgets.Button(description="Artist Popularity across Genres")
artist_popularity_button.on_click(on_artist_popularity_clicked)

top_artists_button = widgets.Button(description="Top 5 Artists in...")
top_artists_button.on_click(on_top_artists_clicked)

back_button = widgets.Button(description="Back to File Loading")
back_button.on_click(on_back_clicked)

# GUI layout
gui_layout = widgets.VBox([
    widgets.HBox([create_genre_stats_button, year_slider]),
    widgets.HBox([artist_popularity_button, artist_text]),
    widgets.HBox([top_artists_button, start_year_slider, end_year_slider]),
    back_button,
    output
])

# Displaying the layout
display(gui_layout) 

# %%
```
